```python
import os
import re
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("1. Loading test datasets...")
test_s1 = pd.read_csv("dataset/test/test_source1.tsv", sep="\t")
test_s2 = pd.read_csv("dataset/test/test_source2.tsv", sep="\t")
test_s3 = pd.read_csv("dataset/test/test_source3.tsv", sep="\t")

# Tag targets
test_s2['target_source'] = 'source2'
test_s3['target_source'] = 'source3'
targets = pd.concat([test_s2, test_s3], ignore_index=True)

# 2. Text normalization
def clean_text(val):
    if pd.isna(val):
        return ""
    val = str(val).lower()
    val = re.sub(r"[^\w\s]", " ", val)
    return " ".join(val.split())

def create_signature(df):
    cols_to_use = [c for c in df.columns if c in ['name', 'address', 'city', 'state', 'postal_code', 'country']]
    if not cols_to_use:
        cols_to_use = [c for c in df.columns if 'id' not in c.lower() and c != 'target_source']
    sigs = df[cols_to_use].fillna("").astype(str).agg(" ".join, axis=1)
    return sigs.apply(clean_text)

print("2. Generating signatures...")
test_s1['signature'] = create_signature(test_s1)
targets['signature'] = create_signature(targets)

s1_id_col = [c for c in test_s1.columns if 'id' in c.lower()][0]
tgt_id_col = [c for c in targets.columns if 'id' in c.lower() and c != 'target_source'][0]

print("3. Vectorizing with sub-word character n-grams...")
vectorizer = TfidfVectorizer(analyzer='char_wb', ngram_range=(3, 4), min_df=2)
vectorizer.fit(pd.concat([test_s1['signature'], targets['signature']]))

s1_vecs = vectorizer.transform(test_s1['signature'])
tgt_vecs = vectorizer.transform(targets['signature'])

# 4. Fast Batch Cosine Similarity & Candidate Generation
print("4. Calculating candidate similarities in batches...")
SIMILARITY_THRESHOLD = 0.72  
TOP_K_CANDIDATES = 15
BATCH_SIZE = 500

candidate_records = []
matching_records = []

num_s1 = s1_vecs.shape[0]
target_ids = targets[tgt_id_col].values
s1_ids = test_s1[s1_id_col].values

for start_idx in range(0, num_s1, BATCH_SIZE):
    end_idx = min(start_idx + BATCH_SIZE, num_s1)
    batch_sims = s1_vecs[start_idx:end_idx].dot(tgt_vecs.T).toarray()
    
    for i, global_s1_idx in enumerate(range(start_idx, end_idx)):
        src1_id = s1_ids[global_s1_idx]
        sims = batch_sims[i]
        
        # Top-K candidate selection
        top_k_indices = np.argpartition(sims, -TOP_K_CANDIDATES)[-TOP_K_CANDIDATES:]
        top_k_indices = top_k_indices[np.argsort(sims[top_k_indices])[::-1]]
        
        matched_ids = []
        for t_idx in top_k_indices:
            score = float(sims[t_idx])
            tgt_id = target_ids[t_idx]
            
            candidate_records.append({
                'source1_id': src1_id,
                'target_id': tgt_id,
                'similarity_score': round(score, 4)
            })
            
            if score >= SIMILARITY_THRESHOLD:
                matched_ids.append(str(tgt_id))
                
        matching_records.append({
            'source1_id': src1_id,
            'matched_target_ids': ",".join(matched_ids) if matched_ids else ""
        })

# 5. Export TSV files
os.makedirs("output", exist_ok=True)

df_candidates = pd.DataFrame(candidate_records)
df_candidates.to_csv("output/candidate_pairs.tsv", sep="\t", index=False)

df_matching = pd.DataFrame(matching_records)
df_matching.to_csv("output/matching_results.tsv", sep="\t", index=False)

print("\nFiles successfully created:")
print(f" - output/candidate_pairs.tsv ({len(df_candidates):,} rows)")
print(f" - output/matching_results.tsv ({len(df_matching):,} rows)")
```

```python
import os
from pathlib import Path

ROOT = Path.cwd()
print(f"Working directory: {ROOT}\n")

# Filename variants to tolerate (official names + the alt names some archives use)
patterns = {
    "train_source1":      ["train_source1.tsv", "trainsource1.tsv"],
    "train_source2":      ["train_source2.tsv", "trainsource2.tsv"],
    "train_source3":      ["train_source3.tsv", "trainsource3.tsv"],
    "train_ground_truth": ["train_ground_truth.tsv", "traingroundtruth.tsv"],
    "test_source1":       ["test_source1.tsv", "testsource1.tsv"],
    "test_source2":       ["test_source2.tsv", "testsource2.tsv"],
    "test_source3":       ["test_source3.tsv", "testsource3.tsv"],
    "validator":          ["validate_submission.py", "validatesubmission.py"],
    "doc_template":       ["Documentation_template.md"],
}

# Single recursive walk, then match filenames (case-insensitive) against it
all_files = [f for f in ROOT.rglob("*") if f.is_file()]

def find_match(name_variants):
    variants_lower = {n.lower() for n in name_variants}
    return [f for f in all_files if f.name.lower() in variants_lower]

discovered = {}

print("=" * 70)
print("FILE DISCOVERY")
print("=" * 70)
for key, variants in patterns.items():
    matches = find_match(variants)
    discovered[key] = matches
    if matches:
        for m in matches:
            size_kb = m.stat().st_size / 1024
            print(f"[FOUND]   {key:20s} -> {m.resolve()}  ({size_kb:.1f} KB)")
    else:
        print(f"[MISSING] {key:20s} -> none of {variants} found under {ROOT}")

missing = [k for k, v in discovered.items() if not v]
extra_matches = {k: v for k, v in discovered.items() if len(v) > 1}

print("\n" + "=" * 70)
if missing:
    print(f"MISSING REQUIRED FILES: {missing}")
    print("STOPPING per fair-play rules — cannot proceed until these are located.")
else:
    print("All required files discovered successfully.")
if extra_matches:
    print(f"NOTE: multiple candidates found for: {list(extra_matches.keys())} "
          f"— resolve ambiguity before Phase 0.2.")
print("=" * 70)

# Create required working directories (idempotent)
for d in ["output", "artifacts", "code/business_entity_resolution/src"]:
    p = ROOT / d
    p.mkdir(parents=True, exist_ok=True)
    print(f"Directory ready: {p.resolve()}")
```
