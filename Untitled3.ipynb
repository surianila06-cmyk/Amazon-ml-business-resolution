```python
import os
import re
import gc
import time
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

DATA_DIR        = "dataset/test"
OUTPUT_DIR      = "output"
SOURCE1_PATH    = os.path.join(DATA_DIR, "test_source1.tsv")
SOURCE2_PATH    = os.path.join(DATA_DIR, "test_source2.tsv")
SOURCE3_PATH    = os.path.join(DATA_DIR, "test_source3.tsv")
CANDIDATES_PATH = os.path.join(OUTPUT_DIR, "candidate_pairs.tsv")
MATCHES_PATH    = os.path.join(OUTPUT_DIR, "matching_results.tsv")

BATCH_SIZE       = 1000
TOP_K             = 15
MATCH_THRESHOLD   = 0.72
LOG_EVERY_N_BATCH = 100

os.makedirs(OUTPUT_DIR, exist_ok=True)

def log(msg):
    print(f"[{time.strftime('%H:%M:%S')}] {msg}", flush=True)

# 1. Load TSVs cleanly
log("Loading TSV datasets...")
read_kw = dict(sep="\t", dtype=str, low_memory=False, keep_default_na=False)
test_s1 = pd.read_csv(SOURCE1_PATH, **read_kw)
test_s2 = pd.read_csv(SOURCE2_PATH, **read_kw)
test_s3 = pd.read_csv(SOURCE3_PATH, **read_kw)

def get_id_col(df):
    for c in df.columns:
        if 'id' in c.lower():
            return c
    return df.columns[0]

s1_id_col = get_id_col(test_s1)
s2_id_col = get_id_col(test_s2)
s3_id_col = get_id_col(test_s3)

# 2. Fast Vectorized Signatures (Word-Level)
def build_sig(df, id_col):
    text_cols = [c for c in df.columns if c != id_col]
    sig = df[text_cols[0]].fillna("").astype(str)
    for col in text_cols[1:]:
        sig = sig + " " + df[col].fillna("").astype(str)
    return (
        sig.str.lower()
        .str.replace(r"[^a-z0-9\s]", " ", regex=True)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

log("Building signatures...")
s1_sig = build_sig(test_s1, s1_id_col)
s1_ids = test_s1[s1_id_col].to_numpy()

s2_sig = build_sig(test_s2, s2_id_col)
s3_sig = build_sig(test_s3, s3_id_col)
target_ids = np.concatenate([test_s2[s2_id_col].to_numpy(), test_s3[s3_id_col].to_numpy()])
target_sig = pd.concat([s2_sig, s3_sig], ignore_index=True)

del test_s1, test_s2, test_s3, s2_sig, s3_sig
gc.collect()

# 3. Memory-Safe Word TF-IDF
log("Fitting TF-IDF Vectorizer (word-level, 1-2 ngrams)...")
vectorizer = TfidfVectorizer(
    analyzer="word",
    ngram_range=(1, 2),
    min_df=5,
    max_features=50_000,
    dtype=np.float32
)
vectorizer.fit(pd.concat([s1_sig, target_sig], ignore_index=True))

log("Transforming matrices...")
s1_matrix = vectorizer.transform(s1_sig).tocsr().astype(np.float32)
targets_T = vectorizer.transform(target_sig).tocsr().astype(np.float32).transpose().tocsc()

del s1_sig, target_sig, vectorizer
gc.collect()

# 4. Batched Matching
n_source1 = s1_matrix.shape[0]
n_targets = targets_T.shape[1]
n_batches = int(np.ceil(n_source1 / BATCH_SIZE))
k_eff = min(TOP_K, n_targets)

log(f"Extracting candidates across {n_batches} batches...")
candidate_rows = []
match_rows = []

for batch_idx in range(n_batches):
    row_start = batch_idx * BATCH_SIZE
    row_end = min(row_start + BATCH_SIZE, n_source1)
    
    # Sparse multiply -> batch dense
    sim_dense = (s1_matrix[row_start:row_end] * targets_T).toarray()
    batch_ids = s1_ids[row_start:row_end]
    
    for local_i in range(sim_dense.shape[0]):
        row = sim_dense[local_i]
        src_id = batch_ids[local_i]
        
        top_idx_unordered = np.argpartition(row, -k_eff)[-k_eff:]
        top_idx = top_idx_unordered[np.argsort(row[top_idx_unordered])[::-1]]
        top_scores = row[top_idx]
        
        matched_targets = []
        for t_idx, score in zip(top_idx, top_scores):
            if score <= 0.0:
                continue
            t_id = target_ids[t_idx]
            candidate_rows.append((src_id, t_id, round(float(score), 4)))
            if score >= MATCH_THRESHOLD:
                matched_targets.append(t_id)
                
        if matched_targets:
            match_rows.append((src_id, ",".join(map(str, matched_targets))))
            
    del sim_dense
    
    if (batch_idx + 1) % LOG_EVERY_N_BATCH == 0 or (batch_idx + 1) == n_batches:
        log(f"Batch {batch_idx + 1}/{n_batches} ({(batch_idx + 1)/n_batches*100:.1f}%) complete.")
        gc.collect()

del s1_matrix, targets_T
gc.collect()

# 5. Output TSVs
log("Writing output TSVs...")
pd.DataFrame(candidate_rows, columns=["source1_id", "target_id", "similarity_score"]).to_csv(CANDIDATES_PATH, sep="\t", index=False)
del candidate_rows
gc.collect()

pd.DataFrame(match_rows, columns=["source1_id", "matched_target_ids"]).to_csv(MATCHES_PATH, sep="\t", index=False)
del match_rows
gc.collect()

log("Output files generated successfully in output/ directory.")
```
